```python
import numpy as np
from collections import Counter

corpus = [
    "이 영화는 정말 재미있고 감동적이다",
    "배우들의 연기가 훌륭하고 영화가 재미있다",
    "영화 스토리가 지루하고 결말이 실망스럽다",
    "감동적인 스토리와 훌륭한 연기가 인상적이다",
    "이 영화는 지루하고 배우 연기도 별로다"
]

tokenized = [ doc.split() for doc in corpus ]
doc_lengths = [ len(tokens) for tokens in tokenized ]

T = sum(doc_lengths)

all_tokens = [ token for doc in tokenized for token in doc ]
V = list( set( [ token for doc in tokenized for token in doc ] ) )
V_size = len(V)

freq = Counter(all_tokens)
print(f"토큰 빈도 : {freq}")

# 빈도 높은 순으로 정렬
for word, count in freq.most_common():
    print(f"{word}: {count}")

TTR = V_size / T

doc_lengths = np.array(doc_lengths)
d_mean = np.mean(doc_lengths)
d_variance = np.var(doc_lengths)
d_std = np.std(doc_lengths)

print(T, V, TTR)
print(d_mean, d_variance, d_std)
```

```python
import re
import unicodedata
from nltk.corpus import stopwords

corpus = [
    "This MOVIE is Really FUN and exciting!",
    "The ACTING is Great, and the Movie is AMAZING.",
    "The Story is BORING, and the Ending is disappointing?",
    "The TOUCHING story and Great ACTING is impressive!",
    "This movie is BORING, and the acting is NOT good."
]

text = [ unicodedata.normalize("NFC", doc) for doc in corpus ]

text = [ doc.lower() for doc in text ]

text = [ re.sub(r"[!,.?]", "", doc) for doc in text ]

tokenized = [ doc.split() for doc in text ]

stop_words = set(stopwords.words('english'))

# tokenized_clean = [token for tokens in tokenized for token in tokens if token not in stop_words]
tokenized_clean = [
    [token for token in tokens if token not in stop_words]
    for tokens in tokenized
]
tokenized_clean
```

```python
import math
from collections import Counter

V = list( set( [ token for tokens in tokenized_clean for token in tokens ] ) )

# TF
def tf(term, doc):
    return doc.count(term)

# DF
def df(term, corpus):
    return sum( [ 1 for doc in corpus if term in doc ] )   

# IDF
def idf(term, corpus):
    return math.log((len(corpus) + 1) / (df(term, corpus) + 1)) + 1

# TF-IDF
def tfidf(term, doc, corpus):
    return tf(term, doc) * idf(term, corpus)

print("=== TF ===")
for vocab in V:
    for i, doc in enumerate(tokenized_clean):
        TF = tf(vocab, doc)
        if TF > 0:
            print(f"TF({vocab}, d{i+1}) = {TF}")
    
print("=== DF & IDF ===")
for vocab in V:
    DF = df(vocab, tokenized_clean)
    IDF = idf(vocab, tokenized_clean)
    print(f"{vocab}: DF={DF}, IDF={IDF:.4f}")

print("=== TF-IDF ===")
for vocab in V:
    for i, doc in enumerate(tokenized_clean):
        tf_idf = tfidf(vocab, doc, tokenized_clean)
        if tf_idf > 0:
            print(f"TF-IDF({vocab}, d{i+1}) = {tf_idf}")
```

```python
import numpy as np

# 문서 벡터 만들기
def make_vector(doc, vocab, corpus):
    return np.array([tfidf(term, doc, corpus) for term in vocab])

def l2_normalize(vector):
    return vector / np.sqrt( sum(vector ** 2) )

def cosine_similarity(a, b):
    return np.dot(a, b) / ( np.linalg.norm(a) * np.linalg.norm(b) )

# 1. 각 문서 벡터 만들기
vectors = [ make_vector(doc, V, tokenized_clean) for doc in tokenized_clean ]

#  2. L2 정규화
vectors_normalized = [l2_normalize(vec) for vec in vectors]

# 3. 모든 문서 쌍 유사도 계산 및 출력
for i in range(len(vectors_normalized)):        # i = 0,1,2,3,4
    for j in range(i+1, len(vectors_normalized)): # j = i+1부터
        sim = cosine_similarity(vectors_normalized[i], vectors_normalized[j])
        print(f"d{i+1} - d{j+1} 유사도: {sim:.4f}")
```
