### 정적 임베딩(Word2Vec/GloVe/FastText)
#### 전이 학습
- 거대 데이터로 미리 학습된 지식을 가져와서 내 문제에 맞게 재사용하는 기법
- Word2Vec, GloVe, FastText는 모두 사전 학습된 단어 벡터를 제공

#### Word2Vec
- 주변 단어로 중심 단어를 예측하거나, 중심 단어로 주변 단어를 예측하면서 단어를 숫자 벡터로 변환
    - CBOW : 주변 단어들 -> 중심 단어 예측
    - Skip-gram : 중심 단어 -> 주변 단어들 예측

#### GloVe
- 전체 말뭉치에서 단어들이 함께 등장하는 빈도(동시 등장 행렬)를 분석해서 벡터를 만듬
    - king - man + woman = queen

#### FastText
- 단어를 subword(n-gram)단위로 쪼개서 학습
- 학습 데이터에 없는 단어(OOV)나 오타도 subword 조각이 겹치면 유사한 벡터를 만들어냄

#### 정적 임베딩의 한계
- 세 모델 모두 정적(Static) 임베딩, 즉 단어마다 벡터가 고정
    - bank -> 항상 같은 벡터 하나
- 문맥에 따라 의미가 달라지는 동음이의어를 처리할 수 없음

```python
# pip install gensim
from gensim.models import Word2Vec

# 토큰화된 문장
sentences = [
    ['I', 'love', 'natural', 'language', 'processing'],
    ['natural', 'language', 'processing', 'is', 'fun'],
    ['I', 'enjoy', 'learning', 'machine', 'learning']
]

# Word2Vec 모델 학습 (CBOW 기반)
w2v_model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, epochs=10000)
# window=2 : 중심 단어 기준 앞뒤 2개 단어를 보겠다
# min_count=1 : 최소 등장 횟수(1) 미만 단어 무시
print(f"language 벡터: {w2v_model.wv['language']}")
# 10차원 숫자 벡터
print(f"natural, language의 유사도: {w2v_model.wv.similarity('language', 'natural')}")
```

```python
# GloVe 모델 로드 + 벡터 연산
import gensim.downloader as api

print('Glove 모델 다운로드 중...대략 66MB')
glove_model = api.load('glove-wiki-gigaword-50')
# 사전 학습된 임베딩 모델을 인터넷에서 자동으로 다운
# glove-wiki-gigaword-50 : Wikipedia + Gigaword 텍스트로 학습한 GloVe 모델
print('로딩완료\n')

print(f"king, queen 유사도: {glove_model.similarity('king', 'queen')}")
print(f"man, woman 유사도: {glove_model.similarity('man', 'woman')}")
```

```python
# 벡터 연산 king - man + woman = queen
result = glove_model.most_similar(positive=['king', 'woman'], negative=['man'], topn=2)
print(f"king-man+woman에 가장 가까운 단어: {result}")
```

```python
# FastText (OOV문제 해결)
from gensim.models import FastText

ft_model = FastText(sentences, vector_size=10, window=2, min_count=1, min_n=3, max_n=6, epochs=10000)
# min_n=3 : subword 최소 글자수
# max_n = 6 subword 최대 글자 수
# 3글자: lan, ang, ngu, gua, uag, age
# 4글자: lang, angu, ngua, guag, uage
# 5글자: langu, angua, nguag, guage
# 6글자: langua, anguag, nguage

# 학습 데이터에 존재하는 단어
print(f"language 벡터: {ft_model.wv['language']}")

# 학습되지 않은 단어 (오타)
print(f"학습되지 않은 단어: languag ({ft_model.wv['languag']})")
# lan, ang, ngu, gua, uag  ← language의 subword와 대부분 겹침!
print(f"languag, language의 유사도: {ft_model.wv.similarity('language', 'languag')}")
```

### ELMo / Bi-LSTM (동적 임베딩)
#### ELMo (Embedding from Language Models)
- 문장 전체를 읽고 문맥에 따라 벡터를 그때그때 생성

#### Bi-LSTM 구조
- ELMo는 Bi-LSTM(양방향)을 여러 층 쌓아서 만듬
    - 순방향 LSTM:  나는 → 밥을 → 먹었다  (왼쪽 → 오른쪽)
    - 역방향 LSTM:  먹었다 → 밥을 → 나는  (오른쪽 → 왼쪽)
- 두 방향을 합치면 특정 단어의 앞 문맥 + 뒤 문맥을 동시에 반영한 벡터가 만들어짐

#### Feature-based 전이학습
- ELMo의 전이학습 방식은 Feature-based
    - 기존에 쓰던 모델 구조를 바꾸지 않음
    - ELMo가 만들어준 동적 벡터를 기존 벡터에 이어붙여(Concatenation) 특징으로 추가
        - 기존 벡터 (10차원) + ELMo 벡터 (24차원) = 합쳐진 벡터 (34차원)

```python
# Bi-LSTM구조 구현
import torch
import torch.nn as nn
import torch.nn.functional as F

embed_size  = 10
hidden_size = 12

elmo_bilstm = nn.LSTM(
    input_size    = embed_size, # 입력 단어 벡터의 차원 = 10
    hidden_size   = hidden_size,# LSTM 내부 은닉 상태 크기 = 12
    num_layers    = 2,          # LSTM을 2겹으로 쌓기
    # 1층 LSTM → 출력 → 2층 LSTM → 최종 출력
    # 층을 쌓을수록 더 추상적인 문맥을 포착 가능
    batch_first   = True,       # 입력 형태를 (배치, 시퀀스, 차원) 순서로
    # batch_first=True 를 쓰는 이유:
    # 기본값은 (시퀀스, 배치, 차원) 순서인데, 대부분의 딥러닝 코드는 (배치, 시퀀스, 차원) 순서를 더 직관적으로 씀
    bidirectional = True        # 순방향 + 역방향 동시 실행
    # 순방향: [나는] → [밥을] → [먹었다]  → 각 위치의 hidden 벡터 생성
    # 역방향: [먹었다] → [밥을] → [나는]  → 각 위치의 hidden 벡터 생성
    # 최종:   두 방향의 벡터를 이어붙임 (concat)
)

# 가상의 데이터 입력 (문장 1개, 5개 단어, 각 단어는 10차원)
dummy_input = torch.rand(1, 5, embed_size)
# 배치 크기 =1, 단어 수 =5, 단어 벡터 차원 = 10
output, (h_n, c_n) = elmo_bilstm(dummy_input)
# (1, 5, 24)
# 배치 = 1, 단어수 = 5, hidden_size(12) * 2방향 = 24
# 모든 시점의 hidden 벡터 (단어마다 하나씩)
# h_n (4, 1, 12) : 마지막 시점의 hidden 벡터
# c_n (4, 1, 12) : 마지막 시점의 cell 벡터(장기 기억)

print(f"입력 텐서 차원: {dummy_input.shape}")
print(f"출력 텐서 차원: {output.shape}")
# 마지막 출력이 24인 이유: hidden_size(12) x 2방향 = 24
```

```python
# 임시 단어사전 구성 (10차원 정적 임베딩)
# 문장 A (은행)  vs  문장 B (강둑)
# → 정적 임베딩: 두 벡터가 완전히 같음 (유사도 = 1.0)
# → ELMo 임베딩: 두 벡터가 달라야 함 (유사도 < 1.0)

word_to_idx = {
    'I':0, 'deposited':1, 'money':2, 'in':3, 'the':4,
    'bank':5, 'walked':6, 'by':7, 'river':8, '.':9
}
static_embedding = nn.Embedding(num_embeddings=10, embedding_dim=embed_size)
# nn.Embedding : 단어 인덱스를 받아서 해당 벡터를 반환하는 조회 테이블
# num_embedding=10 : 단어 사전 크기 (단어 10개)
# embedding_dim=embed_size : 각 단어를 10차원으로 표현

# 문장 A: I deposited money in the bank .
sentence_A_idx = torch.tensor([[word_to_idx[w] for w in ['I', 'deposited', 'money', 'in', 'the', 'bank', '.']]])
# 문장 B: I walked by the river bank .
sentence_B_idx = torch.tensor([[word_to_idx[w] for w in ['I', 'walked', 'by', 'the', 'river', 'bank', '.']]])
# 단어를 인덱스 번호로 바꾼 텐서

# 1. 정적 임베딩 통과 (문맥 반영 안됨)
emb_A = static_embedding(sentence_A_idx)
# emb_A.shape -> (1, 7, 10) 배치=1, 단어=7, 차원=10
# 두 문장 모두 bank의 인덱스는 5 이므로 완전히 동일한 벡터 변환 (문맥을 전혀 보지 않음)
emb_B = static_embedding(sentence_B_idx)

# 2. Bi-LSTM 통과 (양방향 문맥 반영)
elmo_out_A, _ = elmo_bilstm(emb_A)
elmo_out_B, _ = elmo_bilstm(emb_B)
# 같은 위치의 bank라도 서로 다른 벡터가 만들어짐

# 3. bank 위치(인덱스 5)의 벡터 추출
static_bank_A = emb_A[0, 5, :]
static_bank_B = emb_B[0, 5, :]

elmo_bank_A = elmo_out_A[0, 5, :]
elmo_bank_B = elmo_out_B[0, 5, :]

# 4. 코사인 유사도 계산
static_sim = F.cosine_similarity(static_bank_A.unsqueeze(0), static_bank_B.unsqueeze(0))
# unsqeeze(0)가 필요한 이유
# static_bank_A.shape -> (10,) : 1차원 벡터
# F.consine_similarity : 2차원 이상의 텐서를 받음
# unsqueeze(0) : 맨 앞에 차원을 하나 추가 (1, 10)으로 변환
elmo_sim   = F.cosine_similarity(elmo_bank_A.unsqueeze(0),   elmo_bank_B.unsqueeze(0))

print(f"정적 임베딩: {static_sim.item()}")
print(f"ELMO 임베딩: {elmo_sim.item()}")
```

```python
# Feature-based 전이학습 (Concatenation)
# 기존 방식:
# 텍스트 → 정적 임베딩(10차원) → 분류기 → 결과

# Feature-based 전이학습:
# 텍스트 → 정적 임베딩(10차원) ───┐
# 텍스트 → ELMo(24차원)  ────────┤ concat → 합친벡터(34차원) → 분류기 → 결과

original_feature = static_bank_A   # 정적 임베딩 벡터 (10차원)
elmo_feature     = elmo_bank_A     # ELMo 동적 벡터 (24차원)

# Feature-based 전이학습: 기존 벡터에 ELMo 벡터를 이어붙임
combined_feature = torch.cat([original_feature, elmo_feature])
# torch.cat : 텐서를 이어붙이는 함수
# 기본값 dim=0 : 첫 번째 차원 방향으로 이어붙임
# 10차원 + 24차원 = 34차원 벡터 생성

original_feature.shape, elmo_feature.shape, combined_feature.shape
# 결과: (torch.Size([10]), torch.Size([24]), torch.Size([34]))

# 기존 분류기 구조를 전혀 바꾸지 않습니다. 입력 차원만 10 → 34로 늘어남
# Feature-based : 사전학습 모델은 고정, 출력 벡터만 가져다 씀
# Fine-tuning : 사전학습 모델 전체를 내 데이터로 추가 학습
```

### Self-Attention (Transformer 핵심 원리)
#### RNN 계열의 한계
- ELMo에서 사용한 Bi-LSTM을 포함한 모든 RNN계열 모델의 공통 문제점
1. 순차 처리 -> 병렬화 불가
    - RNN : 나는 -> 밥을 -> 먹었다 (반드시 순서대로)
    - GPU의 병렬 연산 능력을 활용할 수 없음
2. 장기 의존성 문제
    - "나는 어제 친구와 함께 맛있는 ???을 먹었다"
    - 문장이 길어질수록 앞의 정보(나는)가 뒤로 갈수록 희미해짐

#### 해결책 : Self-Attention
- Self-Attention은 모든 문장의 모든 단어를 한 번에 행렬 연산으로 처리
    - RNN : 순서대로 1개씩
    - Self-Attention : 나는 밥을 먹었다 (동시에 전부)
- 순서대로 읽을 필요가 없어지므로 GPU 병렬 처리가 가능해짐 => RNN이 퇴출되는 계기

#### Q / K / V
- Self-Attention은 각 단어를 세가지 역할로 변환
1. Query(Q) : 나는 다른 단어들과 얼마나 연관 있는지 => 질문
2. Key(K) :  나는 이런 단어야, 내 정보표는 이거야 => 라벨/지표
3. Value(V) : 나의 실제 의미 정보 => 진짜 값
- bank라는 단어의 경우
    - Q: "나(bank)랑 연관된 단어 누구야?"
    - K: 다른 모든 단어들이 "나는 money야", "나는 river야" 라고 응답
    - V: 실제로 섞어들어갈 각 단어의 의미값

#### Self-Attention 연산 4 단계
1. 점수 계산(Q x K^T) : Q와 K를 내적해서 각 단어 쌍의 연관성 점수를 계산
2. 스케일링(÷ √d_k) : 점수가 너무 커지면 softmax 미분값이 0에 가까워져서 학습이 안됨 -> 차원 크기의 제곱근으로 나눠서 안정화
3. softmax 점수를 확률값(합=1)으로 변환 : 어텐션 가중치
4. 가중합 (Weights x V) : 가중치를 V에 곱하고 더해서 문맥이 반영된 최종 벡터를 만듬

```python
# Q / K / V 행렬 생성
import torch
import torch.nn as nn
import math
import torch.nn.functional as F

torch.manual_seed(42)

# 단어 3개짜리 문장, 임베딩 차원=4, Q/K/V 차원=2
seq_len    = 3
embed_size = 4
d_k        = 2

# 가상의 입력 텐서 (배치=1, 단어수=3, 임베딩=4)
x = torch.rand(1, seq_len, embed_size)
# shape : (1, 3, 4)


# Q/K/V 를 만들기 위한 선형 레이어
W_Q = nn.Linear(embed_size, d_k, bias=False) # 4차원 -> 2차원
W_K = nn.Linear(embed_size, d_k, bias=False) # 4차원 -> 2차원
W_V = nn.Linear(embed_size, d_k, bias=False) # 4차원 -> 2차원

Q = W_Q(x) # shape: (1, 3, 2)
K = W_K(x) # shape: (1, 3, 2)
V = W_V(x) # shape: (1, 3, 2)
# 같은 입력 x 를 세 개의 서로 다른 선형 레이어에 통과

# bias=False 인 이유:
# Self-Attention 수식에서 편향(bias)을 더하지 않는 것이 표준입니다. 어텐션 가중치 계산에서 bias가 방해가 될 수 있기 때문

x.shape, Q.shape, K.shape, V.shape
# 결과: (torch.Size([1,3,4]), torch.Size([1,3,2]), torch.Size([1,3,2]), torch.Size([1,3,2]))

# 차원을 줄이는 이유 (4 -> 2)
# 1. Multi-Head Attention 때문 : Multi-Head에서는 헤드 수만큼 Q/K/V를 만들어서 병렬처리
#    차원을 줄여야 전체 계산량이 유지
# 2. 연산 효율 : 낮은 차원에서 어텐션 점수를 계산하면 더 빠름
```

```python
# 1 & 2. Q K 전치행렬 내적 후 스케일링
scores = (Q @ K.transpose(1, 2)) / math.sqrt(d_k)
# Q  shape: (1, 3, 2)   ← 배치=1, 단어수=3, 차원=2
# K  shape: (1, 3, 2)   ← 그대로는 곱셈 불가
# K.transpose(1, 2) shape : (1, 2, 3) <- 1번 축과 2번축 교환
# Q @ K.T -> (1, 3, 2) x (1, 2, 3) = (1, 3, 3)

#         단어1  단어2  단어3
# 단어1 [  점수   점수   점수  ]  ← 단어1이 다른 단어들을 얼마나 주목하는지
# 단어2 [  점수   점수   점수  ]  ← 단어2가 다른 단어들을 얼마나 주목하는지
# 단어3 [  점수   점수   점수  ]  ← 단어3이 다른 단어들을 얼마나 주목하는지
#  3개 단어가 서로서로에게 주는 연관성 점수 행렬

# 스케일링 (÷ √d_k) 이 필요한 이유:
# d_k 가 커질수록 내적값이 커져서 Softmax 출력이 한쪽으로 쏠림

print(f"step 1&2 K전치 행렬 크기: {K.transpose(1,2).shape}")
print(f"step 1&2 스케일된 어텐션 점수 크기: {scores.shape}")

# 3. softmax로 어텐션 가중치 생성 (합 = 1.0)
atten_weight = F.softmax(scores, dim=-1)
# dim=-1 : 마지막 차원(열 방향)으로 Softmax 적용
print(f"step 3 어텐션 가중치 행렬: {atten_weight}")
print("~~~들의 가중치를 모두 더하면 정확히 1.0이 된다.")

# 4. 가중치와 Value를 곱해 최종 context vector 생성
context_vector = atten_weight @ V
# (1, 3, 3) @ (1, 3, 2) = (1, 3, 2)
print(f"최종 문맥 벡터 크기: {context_vector.shape}")
```

### Multi-Head Attention / Positional Encoding
#### Multi-Head Attention
- Self-Attention을 여러 개의 헤드(Head)로 병렬 실행하는 방식
- 하나로 안되는 이유
    - 단일 Self-Attention → 한 가지 관점으로만 문장 분석
    - 예: "나는 학교에서 친구를 만났다"
    - 관점 1 (문법): 나는 ↔ 만났다   (주어-동사 관계)
    - 관점 2 (의미): 친구 ↔ 만났다   (목적어-동사 관계)
    - 관점 3 (위치): 학교 ↔ 친구     (장소-대상 관계)
    - 하나의 어텐션은 이 중 한 관점밖에 못 잡음, 여러 헤드를 쓰면 동시에 여러 관점을 포착할 수 있음

- 작동 방식
    - 입력 (768차원)
    -    ↓ 헤드 12개로 쪼개기 → 각 헤드 64차원
    -    ↓ 각 헤드가 독립적으로 Self-Attention 수행
    -    ↓ 12개 결과를 다시 이어붙임 (concat)
    - 출력 (768차원)  ← 입력과 같은 차원 유지

#### Positional Encoding
- Self-Attention은 모든 단어를 동시에 처리하기 때문에 단어의 순서 정보가 사라짐
    - "나는 밥을 먹었다"  와  "밥을 나는 먹었다"
    - → Self-Attention 입장에서 수학적으로 동일!
- 이를 해결하기 위해 각 단어 벡터에 위치 정보를 덧셈으로 주입
    - 최종 입력 = 단어 임베딩 + 위치 인코딩
    - 사인/코사인 파동 함수를 사용해서 각 위치마다 고유한 패턴을 만듬

#### Residual Connection + Layer Normalization
- Residual Connection (잔차 연결):
    - 출력 = x + F(x)   ← 블록 통과 전 입력을 그대로 더함
    - 층이 깊어질수록 처음 정보가 희미해지고 기울기 소실이 발생.
    - 원래 입력을 더해주는 고속도로를 만들어서 이를 방지
- Layer Normalization
    - 각 샘플의 차원들을 평균=0, 분산=1로 정규화
    - 값이 너무 커지거나 작아지는 것을 막아 학습을 안정화

#### Transformer 전체 구조
- Encoder
    - 입력 임베딩 + PE 
    - Multi-Head Attention
    - Add & Norm
    - Feed Forward
    - Add & Norm

- Decoder
    - 출력 임베딩 + PE
    - Masked Self-Attention
    - Add & Norm
    - Cross-Attention <- Encoder 출력
    - Add & Norm
    - Feed Forward
    - Add & Norm
- Encoder 만 떼어내서 발전 => BERT
- Decoder 만 떼어내서 발전 => GPT

```python
from transformers import pipeline
# HuggingFace의 pipeline 은 
# 모델 로드 → 토크나이징 → 추론 → 결과 파싱을 한 줄로 처리해주는 고수준 API

unmasker = pipeline('fill-mask', model='bert-base-uncased')
# fill-mask : 빈칸 채우기 태스크
# model='bert-base-uncased : 영어 BERT 기본 모델
# base : 12층, 768차원, 1.1억 파라미터
# uncased : 대소문자 구분 없음 (모두 소문자 처리)
unmasker("Hello I'm a [MASK] model.")
# [MASK] 자리에 들어갈 단어를 확률순으로 반환 
# BERT가 [MASK]위치의 앞뒤 문맥을 동시에 보고 가장 적합한 단어를 예측
# MLM(Maksed Language Model) 사전학습 결과
```

```python
from transformers import BertTokenizer, BertModel

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained(
    'bert-base-uncased',
    output_attentions=True    # ← 어텐션 가중치 반환 옵션
    # 기본값은 False로 어텐션 가중치 반환 X. 
    # True : 모든 레이어의 어텐션 가중치를 같이 반환
)

text = "The cat sat on the mat."
inputs  = tokenizer(text, return_tensors='pt')
outputs = model(**inputs)

tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
# 7개 단어 + [CLS] + [SEP] = 총 9개 토큰
print(f"추출된 토큰들: {tokens}")

all_attentions = outputs.attentions
print(f"전체 레이어 개수: {len(all_attentions)}")
print(f"첫 번째 레이어 어텐션 텐서 크기: {all_attentions[0].shape}")
# all_attentions[0].shape : torch.Size([1, 12, 9, 9])
# 배치, 헤드수 , 토큰수 * 토큰수
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 레이어0, 배치0, 헤드0의 9×9 어텐션 맵 추출
attn_matrix = all_attentions[0][0, 0, :, :].detach().numpy()
print("2D 어텐션 맵 크기:", attn_matrix.shape)

plt.figure(figsize=(8, 6))
sns.heatmap(
    attn_matrix,
    xticklabels=tokens,
    yticklabels=tokens,
    cmap='Blues',
    annot=True,    # 칸 안에 숫자 표시
    fmt=".2f"      # 소수점 2자리
)
plt.title("Self-Attention Map (Layer 0, Head 0)")
plt.xlabel("Key (K)")
plt.ylabel("Query (Q)")
plt.show()
```

### BERT (Encoder 기반 모델)
#### BERT
- Transformer의 Encoder 부분만 수백 층으로 쌓아올린 모델
    - Transformer Encoder만 추출
    - 문장 전체를 병렬로 한 번에 입력
    - 앞뒤 양방향 문맥을 동시에 읽음
    - BERT

#### 두 가지 사전학습 전략
1. MLM(Masked Language Model) - 빈칸 맞추기
    - 입력: "나는 [MASK] 을 먹었다"
    - 학습: 주변 단어로 [MASK] 자리 예측 → "밥"
    - 앞뒤 문맥을 동시에 봐야 맞출 수 있으므로 양방향 문맥 이해를 강제
2. NSP (Next Sentence Prediction) - 다음 문장 맞추기
    - 문장 A: "나는 배가 고팠다."
    - 문장 B: "그래서 밥을 먹었다."  → IsNext (진짜 다음 문장)
    - 문장 C: "하늘이 파랗다."       → NotNext (관계없는 문장)
    - 두 문장의 논리적 연결성을 학습

#### WordPiece Tokenizer + Special Token
- WordPiece - OOV 해결책
    - "unhappiness" → ["un", "##happiness"]
    - "샤갈"        → ["샤", "##갈"]
    - 단어를 최대한 잘게 쪼개서 사전에 없는 단어도 조각으로 처리

- Special Token 3가지
    - [CLS] : (문장 맨 앞) 문장 전체의 핵심 요약 정보를 담음, 분류 태스크에 사용
    - [SEP] : (문장 끝) 문장의 끝ㅇ을 알림. 두 문장 구분에도 사용
    - [MASK] : (중간) 사전학습 시 빈칸 표시

#### 입력 텐서 3대장
- BERT에 입력할 때 3가지 텐서를 함께 넘김
1. input_ids : 토큰을 숫자 ID로 변환
2. Attention_mask : 실제 단어(1) vs 패딩(0) 구분
3. token_type_ids : 문장 A(0) vs 문장 B(1) 구분

- Attention_mask가 필요한 이유
    - 배치 처리를 위해 길이 맞춰야 함
    - 짧은 문장에 [PAD]토큰(0)을 채움
    - [PAD]는 의미 없는 토큰이므로 어텐션 연산에서 제외해야 함
    - attention_mask: 실제 토큰 = 1, 패딩 = 0으로 표시

```python
# 영어 BERT 토크나이저
from transformers import BertTokenizer

tokenizer_en = BertTokenizer.from_pretrained('bert-base-uncased')
# 모델에 맞는 토크나이저 로드, 내부에 약 3만 개의 어휘 사전이 들어있음
text_en = 'I am learning natural language processing'

# 토큰으로 분리
raw_tokens = tokenizer_en.tokenize(text_en)
# 방법 1. tokenize() -> 토큰 리스트만 반환
# ['i', 'am', 'learning', 'natural', 'language', 'processing']
# 방법 2: tokenizer() → input_ids, attention_mask 등 전부 반환
#encoded_dict = tokenizer_en(text_en, add_special_tokens=True)
# 실제 모델에 넣을 때는 방법 2 사용
print(raw_tokens)

# Special Token 추가 후 숫자 ID로 변환
encoded_dict = tokenizer_en(text_en, add_special_tokens=True)
# add_special_tokens=True 가 기본값으로 맨 앞에 [CLS](101), 맨 뒤에 [SEP](102) 가 자동으로 추가
input_ids = encoded_dict['input_ids']
print(f"숫자로 변환된 결과: {input_ids}")

# 숫자 ID를 다시 토큰으로 복원
restored_tokens = tokenizer_en.convert_ids_to_tokens(input_ids)
# # ['[CLS]', 'i', 'am', 'learning', 'natural', 'language', 'processing', '[SEP]']
print(f"숫자를 다시 글자로 복원한 결과: {restored_tokens}")
```

```python
# 배치 처리 + 패딩 + attention_mask
sentences = [
    "I love NLP.",
    "Transformers are incredibly powerful and fast."
]

batch_inputs = tokenizer_en(sentences, padding=True, return_tensors='pt')
for k, v in batch_inputs.items():
    print(k, v)
```

### BERT Fine-tuning(감성 분류)
#### Fine-tuning
- 사전학습된 BERT 전체를 내 데이터로 추가 학습 시키는 방식
    - 사전학습 BERT (수억 개 문장으로 학습된 지식)  
    - 내 태스크 데이터로 추가 학습
    - 태스크에 특화된 모델 완성
- Fine-tuning이 더 강력한 이유는 모델 전체가 태스크에 맞게 가중치가 미세 조정

#### BertForSequenceClassification 구조
- 입력 문장 -> 토크나이징
- [CLS] 토큰들 [SEP] -> BERT Encoder 12 층
- [CLS] 벡터 (768차원) <- 문장 전체 의미 압축 -> Dropout -> Linear
- 분류 결과 (긍정/부정)

#### 학습 흐름
- 입력 + 정답 라벨 -> 모델 forward
- loss 자동 계산 (CrossEntropyLoss) -> loss.backward()
- gradient 계산 -> optimizer.step()
- 가중치 업데이트

HuggingFace 모델은 labels 를 넘기면 loss를 내부에서 자동 계산


```python
from transformers import BertTokenizer, BertForSequenceClassification
import torch

model_name = 'klue/bert-base'
tokenizer  = BertTokenizer.from_pretrained(model_name)
model      = BertForSequenceClassification.from_pretrained(
    model_name,
    num_labels=2    # 긍정/부정 2개 클래스
)
# BertModel 위에 분류 헤드(Linear 레이어)가 추가된 모델

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)
```

```python
from torch.optim import AdamW
from tqdm import tqdm

train_texts = [
    "이 영화 진짜 개꿀잼이네요 ㅋㅋㅋ 완전 강추!",    # 긍정
    "배우들 연기가 훌륭하고 스토리가 감동적입니다.",  # 긍정
    "돈 아깝네요 최악입니다. 절대 보지 마세요.",      # 부정
    "스토리가 너무 지루하고 결말이 어이없음.",        # 부정
]
train_labels = [1, 1, 0, 0]   # 1: 긍정, 0: 부정

# 토크나이징
inputs = tokenizer(
    train_texts,
    padding=True,
    truncation=True,
    return_tensors='pt'
).to(device)

labels    = torch.tensor(train_labels).to(device)
optimizer = AdamW(model.parameters(), lr=5e-5)

model.train()
epochs = 3

for epoch in tqdm(range(epochs)):
    optimizer.zero_grad()
    outputs = model(**inputs, labels=labels)
    loss    = outputs.loss
    loss.backward()
    optimizer.step()
    print(f"epoch: {epoch+1} / {epochs} loss: {loss.item()}")
```

```python
model.eval()

test_texts = [
    "시간 가는 줄 모르고 봤습니다. 인생작 등극!",
    "도대체 무슨 내용을 말하고 싶은 건지 모르겠네요. 돈버림."
]

test_inputs = tokenizer(
    test_texts,
    padding=True,
    truncation=True,
    return_tensors='pt'
).to(device)

with torch.no_grad():
    outputs     = model(**test_inputs)
    logits      = outputs.logits
    predictions = torch.argmax(logits, dim=-1)
    print(predictions)
# 결과: tensor([1, 0])  ← 긍정, 부정
```

### GPT (Decoder 기반 텍스트 생성)
#### BERT vs GPT 근본 차이
- BERT (Encoder) : 
    - 문장 전체를 한 번에 통째로 봄
    - "나는 [MASK] 을 먹었다" -> 앞뒤 동시에 참고
    - => 독해 / 분류에 최적화
- GPT (Decoder) :
    - 오직 왼쪽(과거)만 봄
    - "나는 밥을" -> 다음 단어 "먹었다" 예측
    - => 텍스트 생성에 최적화

#### Masked Self-Attention (치팅 방지)
- GPT는 학습할 때 문장 전체를 한 번에 입력받음
- 그런데 다음 단어를 예측해야 하는데 옆에 정답이 있으면 컨닝(cheating)이 됨
- 이를 방지하기 위해 미래 위치를 마스킹
- Softmax 통과 시 마스킹된 위치의 확률이 0이 되도록 처리

#### 자기회귀 (Autoregressive) 생성
- 이전에 생성한 단어를 다시 입력으로 사용하는 방식

#### Decoding 전략
1. Greedy : 항상 확률 1등 단어 선택
    - 빠르지만 반복 문장 생성
2. Sampling : 확률에 따라 랜덤 선택
    - 다양하지만 가끔 엉뚱함
3. Top-k : 상위 k개중 샘플링
    - 균형 잡힌 다양성
4. top-p : 누적 확률 p 안에서 샘플링
    - 더 자연스러운 문장

```python
# KoGPT2 모델 로드
import torch
from transformers import GPT2LMHeadModel, PreTrainedTokenizerFast

tokenizer = PreTrainedTokenizerFast.from_pretrained(
    'skt/kogpt2-base-v2',
    bos_token='</s>',    # 문장 시작 토큰
    eos_token='</s>',    # 문장 끝 토큰
    unk_token='<unk>',   # 미등록 단어 토큰
    pad_token='<pad>',   # 패딩 토큰
    mask_token='<mask>'  # 마스크 토큰
)

model = GPT2LMHeadModel.from_pretrained('skt/kogpt2-base-v2')
```

```python
# Greedy Decoding (앵무새 병)
prompt_text = "인공지능이 인간을 대체할 수 있을까?"
input_ids   = tokenizer.encode(prompt_text, return_tensors='pt')

with torch.no_grad():
    greedy_output = model.generate(
        input_ids,
        max_length    = 50,
        pad_token_id  = tokenizer.pad_token_id
    )

generated_text = tokenizer.decode(greedy_output[0], skip_special_tokens=True)
print(generated_text)
```

```python
# Sampling Decoding (창의적 생성)
with torch.no_grad():
    sample_output = model.generate(
        input_ids,
        max_length         = 50,
        do_sample          = True,    # 주사위 굴리기
        top_k              = 50,      # 상위 50개 중 샘플링
        top_p              = 0.9,     # 누적 확률 90% 안에서
        temperature        = 0.5,     # 창의성 조절
        repetition_penalty = 1.2,     # 반복 단어 벌점
        pad_token_id       = tokenizer.pad_token_id
    )

creative_text = tokenizer.decode(sample_output[0], skip_special_tokens=True)
print(creative_text)
```

```python
# GPT2 Fine-tuning (Q&A 형식)
from torch.optim import AdamW

train_text = [
    '질문 : 자연어 처리가 뭔가요? 답변 : 컴퓨터가 인간의 언어를 이해하고 분석하는 인공지능 기술입니다.',
    '질문 : 오늘 기분이 어때? 답변 : 저는 인공지능이라 기분을 느낄 수 없지만, 당신과 대화해서 기뻐요.'
]

inputs       = tokenizer(train_text, padding=True, truncation=True, return_tensors='pt')
input_ids    = inputs['input_ids']
attention_mask = inputs['attention_mask']

optimizer = AdamW(model.parameters(), lr=5e-5)
model.train()
epochs = 10

for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=input_ids)
    loss    = outputs.loss
    loss.backward()
    optimizer.step()
    print(f"epoch {epoch+1} / {epochs} loss = {loss.item()}")
```

```python
# 다른 질문으로 추론 (한계 확인)
model.eval()
test_prompt = '질문: 철수가 영희를 괴롭히면 어떻게 할까요? 답변:'
test_ids    = tokenizer.encode(test_prompt, return_tensors='pt')

with torch.no_grad():
    test_output = model.generate(
        test_ids,
        max_length   = 40,
        temperature  = 0.8,
        pad_token_id = tokenizer.pad_token_id
    )
print(tokenizer.decode(test_output[0], skip_special_tokens=True))
# 출력: 질문: 철수가 영희를... 답변: 저는 인공지능이라 기분을 느낄 수 없지만...
```
