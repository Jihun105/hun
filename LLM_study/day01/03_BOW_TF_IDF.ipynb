```python
import math

# 예제 코퍼스
corpus = [
    ["나는", "자연어", "처리를", "공부한다"],
    ["자연어", "처리는", "재미있다"],
    ["나는", "자연어를", "좋아하고", "자연어", "처리를", "실습한다"]
]

N = len(corpus)

# 어휘 집합 
vocab = list( set( token for doc in corpus for token in doc ) )

# TF 계산
def tf(term, doc):
    return doc.count(term)

# DF 계산
def df(term, corpus):
    return sum( 1 for doc in corpus if term in doc )

# IDF 계산
def idf(term, corpus):
    return math.log( (N + 1) / (df(term, corpus) + 1) ) + 1

# TF-IDF 계산
def tfidf(term, doc, corpus):
    return tf(term, doc) * idf(term, corpus)

# 결과 출력
print("=== TF ===")
for term in vocab:
    for i, doc in enumerate(corpus):
        val = tf(term, doc)
        if val > 0:
            print(f"TF({term}, d{i+1} = {val}")

print("\n=== DF & IDF ===")
for term in vocab:
    d = df(term, corpus)
    i = idf(term, corpus)
    print(f"{term}: DF={d}, IDF={i:.4f}")

print("\n=== TF-IDF ===")
for term in vocab:
    for i, doc in enumerate(corpus):
        val = tfidf(term, doc, corpus)
        if val != 0:
            print(f"TF-IDF({term}, d{i+1}) = {val:.4f}")
```
