```python
corpus = [
    "나는 자연어 처리를 공부한다",
    "자연어 처리는 재미있다",
    "나는 데이터 분석과 자연어 처리를 함께 배운다"
]
```

```python
# 토큰화 (공백 기준)
tokenized = [ doc.split() for doc in corpus ]

# 문서 길이 (document length)
doc_lenghts = [ len(doc) for doc in tokenized ]

# 전체 토큰 수 (token count)
T = sum(doc_lenghts)

# 어휘 집합 (vocabulary)
all_tokens = [ token for doc in tokenized for token in doc ]
V = set(all_tokens)

# TTR
TTR = len(V) / T # 중복 제거 고유 토큰 수 / 전체 토큰 수

print(f"문서 길이 : {doc_lenghts}")
print(f"전체 토큰 수 T : {T}")
print(f"어휘 집합 V : {V}")
print(f"어휘 크기 |V| : {len(V)}")
print(f"TTR : {len(V)} /{T} = {TTR:.3f}")
```

```python
from collections import Counter

# 빈도 계산
freq = Counter(all_tokens)
print(f"토큰 빈도 : {freq}")

# 빈도 높은 순으로 정렬
for word, count in freq.most_common():
    print(f"{word}: {count}")
```

```python
import numpy as np

lengths = np.array(doc_lenghts) # 수학 연산을 하기위해 array 변경(np를 쓰기 위해)

mean = np.mean(lengths)
variance = np.var(lengths)
std = np.std(lengths)

print(f"평균 문서 길이 : {mean:.2f}")
print(f"분산 : {variance:.4f}")
print(f"표준편차 : {std:.4f}")
```

$$\sigma^2 = \frac{1}{N}\sum_{i=1}^{N}(\ell(d_i) - \mu)^2$$

- $\sigma^2$ : 분산
- $N$ : 문서 수
- $\ell(d_i)$ : i번째 문서 길이
- $\mu$ : 평균 문서 길이
- $\sum_{i=1}^{N}$ : i=1부터 N까지 합산
