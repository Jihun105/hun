```python
import numpy as np

# 예제 벡터
u = np.array([3, 4])
v = np.array([1, 2])

# L1 정규화
u_l1 = u / np.sum(np.abs(u))
v_l1 = v / np.sum(np.abs(v))

# L2 정규화
u_l2 = u / np.linalg.norm(u)
v_l2 = v / np.linalg.norm(v)

# 최대값 정규화
u_max = u / np.max(u)
v_max = v / np.max(v)

print("=== 정규화 결과 ===")
print(f"L1   - u' : {u_l1}, v' : {v_l1}")
print(f"L2   - u' : {u_l2}, v' : {v_l2}")
print(f"Max  - u' : {u_max}, v' : {v_max}")

# 코사인 유사도 
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("\n=== 코사인 유사도 ===")
print(f"정규화 전 : {cosine_similarity(u,v):.4f}")
print(f"L2 정규화 후 : {cosine_similarity(u_l2,v_l2):.4f}")
```
