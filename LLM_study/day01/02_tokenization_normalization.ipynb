```python
import re
import unicodedata

# 원문
text = "오늘은 NLP를 공부한다! NLP는 정말 재미있다. NLP, NLP!"

# 불용어 정의
stopwords = {"는", "을", "를", "이", "가", "은", "도"}

# 1. 유니코드 정규화
text = unicodedata.normalize("NFC", text)

# 2. 소문자화
text = text.lower()

# 3. 구두점 제거
# text = re.sub(r"[^\w\d]", "", text) # 한국어에서 공백도 같이 날아가버림
text = re.sub(r"[!.,?]", "", text)

# 4. 토큰화
tokens = text.split()

# 5. 불용어 제거
tokens_clean = [ t for t in tokens if t not in stopwords ] 

print(f"정규화 후 원문 : {text}")
print(f"토큰화 결과 : {tokens}")
print(f"불용어 제거 후 : {tokens_clean}")

# re.sub(패턴, 바꿀것, 텍스트)패턴에 맞는 부분 교체! → ""
# re.findall(패턴, 텍스트)패턴에 맞는 부분 찾기모든 숫자 찾기
# re.split(패턴, 텍스트)패턴 기준으로 쪼개기공백/쉼표 기준 split
```
