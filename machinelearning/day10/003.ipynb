```python
torch.manual_seed(42)
class SigmoidNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(4, 8)
        self.layer2 = nn.Linear(8, 8)
        self.layer3 = nn.Linear(8, 8)
        self.layer4 = nn.Linear(8, 8)
        self.layer5 = nn.Linear(8, 1)
    
    def forward(self, x):
        x = torch.sigmoid(self.layer1(x))
        x = torch.sigmoid(self.layer2(x))
        x = torch.sigmoid(self.layer3(x))
        x = torch.sigmoid(self.layer4(x))
        x = self.layer5(x)
        return x

sigmoid_net = SigmoidNet()
sigmoid_net
```

```python
x_input = torch.randn(1,4)
y_target = torch.tensor([[1.0]])
output = sigmoid_net(x_input) # forward가 호출됨
loss = (y_target - output)**2
loss.backward()
```

```python
torch.manual_seed(42)
class ReLUNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(4, 8)
        self.layer2 = nn.Linear(8, 8)
        self.layer3 = nn.Linear(8, 8)
        self.layer4 = nn.Linear(8, 8)
        self.layer5 = nn.Linear(8, 1)
    
    def forward(self, x):
        x = torch.relu(self.layer1(x))
        x = torch.relu(self.layer2(x))
        x = torch.relu(self.layer3(x))
        x = torch.relu(self.layer4(x))
        x = self.layer5(x)
        return x

relu_net = ReLUNet()
relu_net

x_input = torch.randn(1,4)
y_target = torch.tensor([[1.0]])
output = relu_net(x_input) # forward가 호출됨
loss = (y_target - output)**2
loss.backward()

for name, param in relu_net.named_parameters():
    grad_magnitude = param.grad.abs().mean().item()
    status = "소실" if grad_magnitude < 0.001 else "보통" if grad_magnitude < 0.01 else "양호"
    print(name, grad_magnitude, status)
    
```

```python
torch.manual_seed(42)
class ReLUNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(4, 8)
        self.layer2 = nn.Linear(8, 8)
        self.layer3 = nn.Linear(8, 8)
        self.layer4 = nn.Linear(8, 8)
        self.layer5 = nn.Linear(8, 1)
        self.leakyReLU = nn.LeakyReLU()
    
    def forward(self, x):
        x = torch.leakyReLU(self.layer1(x))
        x = torch.leakyReLU(self.layer2(x))
        x = torch.leakyReLU(self.layer3(x))
        x = torch.leakyReLU(self.layer4(x))
        x = self.layer5(x)
        return x

relu_net = ReLUNet()
relu_net

x_input = torch.randn(1,4)
y_target = torch.tensor([[1.0]])
output = relu_net(x_input) # forward가 호출됨
loss = (y_target - output)**2
loss.backward()

for name, param in relu_net.named_parameters():
    grad_magnitude = param.grad.abs().mean().item()
    status = "소실" if grad_magnitude < 0.001 else "보통" if grad_magnitude < 0.01 else "양호"
    print(name, grad_magnitude, status)
    
```
